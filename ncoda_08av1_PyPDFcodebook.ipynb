````python
import os
import pandas as pd
import importlib.util
````

````python
import pypdfcodebook
print(pypdfcodebook.__version__)
````

````python
from pypdfcodebook.pdfcb_03c_codebook import codebook
from pyncoda.ncoda_00f_hua_structure_v2 import adjust_hua_datastructure_baseyear
````

````python
# Paths to codebook files
projectoverview_path = os.path.join('pyncoda', 'ncoda_00a_projectoverview.md')
keyterms_path = os.path.join('pyncoda', 'CommunitySourceData\\api_census_gov\\acg_00a_keyterms.md')
# datastructure will be made with function call below
# datastructure_path = os.path.join('pyncoda', 'ncoda_00f_hua_structure.py') 
csv_path = os.path.join('OutputData', 'hua_v2-1-0_Seaside_OR_NSI_2020_rs9885_NSI.csv')

footer_image_path = os.path.join('IN-CORE_HRRC_Banner.png')
````

````python
# Load CSV
input_df = pd.read_csv(csv_path)
input_df.head()
````

````python
hua_DataStructure = adjust_hua_datastructure_baseyear(baseyear = '2020', state_county = '41007')
````

````python
hua_DataStructure
````

````python
# Set up output in tests directory
output_filename = "pypdfcodebook_test"  # Just the name without extension
outputfolders = {'top' : 'OutputData'}
````

````python
# Create codebook
pdfcodebook = codebook(
    input_df=input_df,
    header_title='Housing Unit Allocation Codebook',
    datastructure=hua_DataStructure,
    projectoverview=projectoverview_path,
    keyterms=keyterms_path,
    output_filename=output_filename,
    outputfolders=outputfolders,
    footer_image_path=footer_image_path
)
pdfcodebook.create_codebook()
````
